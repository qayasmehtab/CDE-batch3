# Dictionaries Looping value

```python
bio_data = {
    "first_name": "Qayas",
    "last_name": "Mehtab",
    "age": 35,
    "address": "Malir",
    "Id": "105"
}

# Loop through keys only & Value 
for key in bio_data:
    print(key)
for value in bio_data:
    print(value)
for key, value in bio_data.items():
    print(f"{key} : {value}")

marks = {"Ali": 85, "Sara": 92, "Ahmed": 78}

# Print report
for name, score in marks.items():
    print(f"{name} scored {score} marks")
```
