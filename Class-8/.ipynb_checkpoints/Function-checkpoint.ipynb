```python
def login(email, password):
    if email in user_info.keys():
        if password == user_info[email]["password"]:
            print(f"{user_info[email]['name']}, Login successfully.")
        else:
            print("Wrong password.")
    else:
        print("Email not found.")
```

```python
login("qayasabbasigmail.com", 1234)
login("qayasabbasigmail.com", 1214)
login("aayasabbasigmail.com", 1234)
```

# Task:
* Create a simple calculator  with features like addition, subtraction, multiplication, and division 

```python
def calculator():
    fnum = int(input("Enter first number: "))
    snum = int(input("Enter second number: "))
    opt = input("Enter operation (+,-,*,/): ")

    if opt == "+":
        print(fnum + snum)
    elif opt == "-":
        print(fnum - snum)
    elif opt == "*":
        print(fnum * snum)
    elif opt == "/":
        if snum == 0:
            print("Cannot divide by zero.")
        else:
            print(fnum / snum)
    else:
        print("Invalid operator.")       
```

```python
calculator()
```

```python
def Gradesystem(marks):
    total = 500
    obt_per = marks / total * 100

    if obt_per >= 90:
        return "A+"
    elif obt_per >= 80:
        return "A"
    elif obt_per >= 70:
        return "B"
    elif obt_per >= 60:
        return "C"
    elif obt_per >= 50:
        return "D"
    else:
        return "Fail"
```

```python
Gradesystem(400)
```
