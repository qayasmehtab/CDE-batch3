# CSV file Handling

```python
import csv
```

```python
with open("csv_test.csv", "w") as f:
    csv_data = csv.writer(f)
    csv_data.writerow(["S:No","Name", "Father.Name"])
```

```python
with open("csv_test.csv", "r")as f:
    csv_data = csv.reader(f)
    for i in csv_data:
        print(i)
```

```python
import csv

new_data = [
    ["Name", "Age", "City",],
    ["Ali", 25, "Karachi"],
    ["Sara", 30, "Lahore"]
]

with open("csv_test.csv", "w") as f:
    writer = csv.writer(f)
    writer.writerows(new_data)

print("update file.")
```

```python
with open("csv_test.csv", "a") as f:
    writer = csv.writer(f)
    writer.writerow(["Qayas", 28, "Islamabad"])
```

```python
with open("csv_test.csv", "r")as f:
    csv_data = csv.reader(f)
    for i in csv_data:
        print(i)
```

# Json File

```python
import json
```

```python
with open("json_test.json", "w") as f:
    json_content = json.dump({
        "name": "John Doe",
        "age": 30,
        "email": "john.doe@example.com",
        "isMember": True,
        "skills": ["JavaScript", "Python", "HTML"],
        "address": {
            "street": "123 Main St",
            "city": "New York",
            "zip": "10001"
        }
    }, f)
```

```python
with open("json_test.json", "r") as f:
    json_content = json.load(f)
    json_content["weight"] = 70
```

```python
json_content
```

```python
with open("json_test.json", "w")as f:
    json.dump(json_content, f)
```

```python
with open ("json_test.json", "r") as f:
    json_content = json.load(f)
    print(json_content)
```

# Error Handling

```python
print("code Begin")
6/0
print("code End")
```

```python
try:
    6 / 0
except Exception as e:
    print("Number cannot be divided by 0.", e)

print("Code End")
```

```python
try:
    with open("demo.txt", "r")as f:
        print(f.read())
except FileNotFoundError as e:
    print("File not found.", e)

print("Code End")
```
