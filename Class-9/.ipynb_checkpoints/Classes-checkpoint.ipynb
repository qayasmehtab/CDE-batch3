* 1 Inheritance ()
* 2 Abstrcation

```python
class Parent:
    def __init__(self, eye_color, hair_color):   
        self.eye_color = eye_color
        self.hair_color = hair_color

    def drive(self):
        print("Parent can drive")
```

```python
par_obj = Parent("Brown", "Brown")

print(par_obj.eye_color)   # Output: Brown
print(par_obj.hair_color)
par_obj.drive()
```

```python
class Child(Parent):
    def __init__(self, eye_color, hair_color, skin_color):
        super().__init__(eye_color, hair_color)
        self.skin_color = skin_color
```

```python
chi_obj = Child("Brown", "Brown", "White")
```

```python
print(chi_obj.eye_color)   # Brown
print(chi_obj.hair_color)  # Brown
print(chi_obj.skin_color)  # White

chi_obj.drive()            # Parent can drive
```

```python
class Father:
    def __init__(self, name, language):
        self.name = name
        self.language = language

    def drive(self):
        print('Father can drive.')

class Mother:
    def __init__(self, language):
        self.language = language

    def cook(self):
        print('Mother can cook.')

class Child(Father, Mother):
    def __init__(self, name, language):
        Father.__init__(self, name, language)
        Mother.__init__(self, language)
        self.name = name
        self.language = language

    def programmer(self):
        print('Child can write a program.')
```

```python
f_obj = Father('Mehmood', 'Sindi')

print(f_obj.name)
print(f_obj.language)
f_obj.drive()
```

```python
m_obj = Mother('Punjabi')

print(m_obj.language)
m_obj.cook()
```

```python
f_obj.language
```

```python
c_obj = Child('Ali', 'Punjabi')

print(c_obj.name)
print(c_obj.language)
c_obj.programmer()
```

# OverLoading

```python
from typing import overload

class Adder():

    @overload
    def add (self, x:int , y:int) -> int:
        ...

    @overload
    def add (self, x:float , y:float) -> float:
        ...

    @overload
    def add (self, x:str , y:str) -> str:
        ...

    # ACTUAL FUNCTIONALITY
    def add (self, x, y):
        if isinstance(x, int) and isinstance(y, int):
            return x+y
        elif isinstance(x, float) and isinstance(y, float):
            return x+y
        elif isinstance(x, str) and isinstance(y, str):
            return x+y
        else: 
            print("Type Error")
```

```python
add = Adder()
```

```python
add.add(3, 6)
```

```python
add.add("Umair", "Nawaz")
```

### Abstractuib
* The concept of reducing complexity of a calss 

```python
class payment:
    def pay(self):
        pass

class CreditCard(payment):
    def pay(self,amount):
        print(f"paying {amount} using Credit Card .")

class DebitCard(payment):
    def pay(self,amount):
        print(f"paying {amount} using DebitCard .")
        

class JazzCash(payment):
    def pay(self,amount):
        print(f"paying {amount} using JazzCash .")
        
class Easypaisa(payment):
    def pay(self,amount):
        print(f"paying {amount} using Easypaisa .")
```

```python
cred_obj = CreditCard()
Debit_obj = DebitCard()
Easy_obj = Easypaisa()
jazz_obj = JazzCash()
```

```python
cred_obj.pay(23)     
Debit_obj.pay(50)
Easy_obj.pay(75)
jazz_obj.pay(10)
```

### Task:
Create an abrtact calass Notification:
* It should have an abstract method **Send(massage)** that accepts a message as a parameter.
* It should also have a regular method **Show_info()** that prints "Sending Notification."
* Create two class that inherit from notification
* 1 Email Notification ---> should implement send() to print "sending Email:"
* 2 SMS Notification ----> should implement send () to print "Sending SMS:"

```python
class notification:
    def send(self, massage):
        pass


    def show_info(self):
        print("sending notification.")

        

class Emailnotification(notification):
    def send(self, massage):
        print(f"sending Email:{massage}")

class SMSnotification(notification):
    def send(self, massage):
        print(f"sending SMS:{massage}")


Email = Emailnotification()
SMS = SMSnotification()

Email.show_info()
Email.send("Your trancastion has been succesfull")

SMS.show_info()
SMS.send("Your trancastion NUmber 2343567")
```

```python
class Notification():
    def send(self, message):
        pass
    
    def show_info(self):
        print("Sending Notification.")


class EmailNotification(Notification):
    def send(self, message):
        print(f"Sending Email: {message}")


class SMSNotification(Notification):
    def send(self, message):
        print(f"Sending SMS: {message}")


email = EmailNotification()
sms = SMSNotification()

email.show_info()
email.send("Welcome to our platform!")

sms.show_info()
sms.send("Your verification code is 123456")
```

```python
class StudentLogin:
    def __init__(self, username, password):
        self.__username = username
        self.__password = password
    
    def __db_connection(self, input_username, input_password):
        if input_username == self.__username and input_password == self.__password:
            return True
        return False
    
    def login(self, username, password):
        if self.__db_connection(username, password):
            print("Valid user. Login successful!")
        else:
            print("Invalid user.")


student = StudentLogin("admin", "12345")

student.login("admin", "12345")
student.login("admin", "wrong")
```
